```python
#write a program to replace Last vale of tuple in a list to 100
numbers = [(10,20,40), (40,50,60), (70, 80,90)]
A= []
for tup in numbers:
    new_tup = tup[:-1] + (100,)
    A.append (new_tup)
print("Original List:", numbers)
print("updated List: ", A)
```

```python
#python program to find the maximum and minimum element in a tuple
A= (25,10,45, 30,60)
maximum = max(A)
minimum = min(A)
print("tuple:", A)
print("Maximum number:", maximum)
print("Minimum number:", minimum)
```
